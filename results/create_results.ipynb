```python
import matplotlib.pyplot as plt

# Dữ liệu thực tế với tổng 4283 truy vấn
categories = [
    "Nhận diện hành vi\nvi phạm",
    "Tình huống\ncụ thể",
    "Suy luận ra chế tài,\nmức phạt hoặc căn cứ"
]
counts = [1567, 1789, 927]  # Tổng = 4283

# Vẽ biểu đồ
plt.figure(figsize=(8, 5))
bars = plt.bar(categories, counts, color='#4a90e2')

# Thêm nhãn giá trị trên từng cột
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 25, f"{yval}", ha='center', va='bottom', fontsize=10)

# Cấu hình biểu đồ
plt.xlabel("Loại câu hỏi", fontsize=15)
plt.ylabel("Số lượng câu hỏi", fontsize=15)
plt.ylim(0, max(counts) * 1.2)
plt.tight_layout()

# Hiển thị
plt.show()
```
